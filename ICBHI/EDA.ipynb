# Exploratory data analysis

## 1. Import libraries

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.figure_factory as ff
```

## 2. Load metadata

```python
df_all = pd.read_csv("metadata.csv")
display(df_all)
df = pd.read_csv("fragments_metadata.csv")
display(df)
df_patients = pd.read_csv("ICBHI/ICBHI_Challenge_demographic_information.txt", sep="\t", header=None, names=["patient_id", "age", "sex", "BMI", "weight", "height"])
df_patients
```

## 3. EDA

### 3.1. Dataset analysis

```python
n_records = len(df_all)
n_segments = len(df)
n_patients = len(df["patient_id"].unique())

tot = sum(df_all["duration"])
mins_t, s_t = int(tot//60), round(tot%60, 2)
h_t, mins_t = mins_t//60, mins_t%60

dur = sum(df["duration"])
mins, s = int(dur//60), round(dur%60, 2)
h, mins = mins//60, mins%60

print(f"Number of records: {n_records}")
print(f"Number of segments: {n_segments}")
print(f"Number of patients: {n_patients}")
print(f"Total recorded time: {h_t} h {mins_t} min {s_t} s")
print(f"Corpus duration: {h} h {mins} min {s} s ({dur/tot*100:.2f}%)")
```

### 3.2. Filter by age

```python
df_all_merge = df_all.merge(df_patients[["patient_id", "age"]], on="patient_id", how="left")
df_all_filtered = df_all_merge[df_all_merge["age"] <= 21].reset_index(drop=True)
df_all_filtered
```

```python
df_merge = df.merge(df_patients[["patient_id", "age"]], on="patient_id", how="left")
df_filtered = df_merge[df_merge["age"] <= 21].reset_index(drop=True)
df_filtered
```

#### 3.2.1. Dataset analysis after filter

```python
n_records_filtered = len(df_all_filtered)
n_segments_filtered = len(df_filtered)
n_patients_filtered = len(df_filtered["patient_id"].unique())

tot_filtered = sum(df_all_filtered["duration"])
mins_t, s_t = int(tot_filtered//60), round(tot_filtered%60, 2)
h_t, mins_t = mins_t//60, mins_t%60

dur_filtered = sum(df_filtered["duration"])
mins, s = int(dur_filtered//60), round(dur_filtered%60, 2)
h, mins = mins//60, mins%60

print(f"Number of records: {n_records_filtered} ({n_records_filtered/n_records*100:.2f}%)")
print(f"Number of segments: {n_segments_filtered} ({n_segments_filtered/n_segments*100:.2f}%)")
print(f"Number of patients: {n_patients_filtered} ({n_patients_filtered/n_patients*100:.2f}%)")
print(f"Total recorded time: {h_t} h {mins_t} min {s_t} s ({tot_filtered/tot*100:.2f}%)")
print(f"Corpus duration: {h} h {mins} min {s} s ({dur_filtered/dur*100:.2f}%)")
```

### 3.3. Age distribution

```python
# --- 1. Datos ---
ages = df_patients["age"].dropna()
ages_filtered = ages[ages <=21]

# --- 3. Crear figura: 2 filas, eje X compartido ---
fig = make_subplots(
    rows=3, cols=1,
    shared_xaxes=True,
    row_heights=[0.75, 0.25, 0.25],   # histograma más grande, boxplot como "resumen" fino
    vertical_spacing=0.03,
)

# --- Histograma (normalizado a densidad para que la escala coincida con la KDE) ---
fig.add_trace(
    go.Histogram(
        x=ages,
        name="Duration",
        marker_color="steelblue",
        opacity=0.75,
        showlegend=False,
    ),
    row=1, col=1
)
fig.add_trace(
    go.Histogram(
        x=ages_filtered,
        name="Duration",
        marker_color="forestgreen",
        opacity=0.75,
        showlegend=False,
    ),
    row=1, col=1
)

# --- Boxplot horizontal ---
fig.add_trace(
    go.Box(
        x=ages,
        name="Age",
        marker_color="steelblue",
        line_color="steelblue",
        boxpoints="outliers",   # muestra outliers como puntos
        showlegend=False,
        boxmean="sd"
    ),
    row=2, col=1
)

# --- Boxplot horizontal ---
fig.add_trace(
    go.Box(
        x=ages_filtered,
        name="Age (≤ 21 y/o)",
        marker_color="forestgreen",
        line_color="forestgreen",
        boxpoints="outliers",   # muestra outliers como puntos
        showlegend=False,
        boxmean="sd"
    ),
    row=3, col=1
)

# --- 4. Layout ---
fig.update_layout(
    template="plotly_white",
    title_text="Age patient distribution",
    font=dict(size=15),
    title_font=dict(size=20),
    bargap=0.02,
    barmode="overlay"
)
fig.update_yaxes(title_text="Count", row=1, col=1)
fig.update_yaxes(showticklabels=True, row=3, col=1)
fig.update_xaxes(title_text="Age (years)", row=3, col=1)

fig.show()
fig.write_image("assets/age_patient_distribution.png", scale=5)
```

### 3.4. Duration distribution

```python
# --- 1. Datos ---
durations = df["duration"].dropna()
durations_filtered = df_filtered["duration"].dropna()


# --- 3. Crear figura: 2 filas, eje X compartido ---
fig = make_subplots(
    rows=3, cols=1,
    shared_xaxes=True,
    row_heights=[0.75, 0.25, 0.25],   # histograma más grande, boxplot como "resumen" fino
    vertical_spacing=0.03,
)

# --- Histograma (normalizado a densidad para que la escala coincida con la KDE) ---
fig.add_trace(
    go.Histogram(
        x=durations,
        name="Duration",
        marker_color="chocolate",
        opacity=0.75,
        showlegend=False,
    ),
    row=1, col=1
)

fig.add_trace(
    go.Histogram(
        x=durations_filtered,
        name="Duration",
        marker_color="forestgreen",
        opacity=0.75,
        showlegend=False,
    ),
    row=1, col=1
)

# --- Boxplot horizontal ---
fig.add_trace(
    go.Box(
        x=durations,
        name="Duration",
        marker_color="chocolate",
        line_color="chocolate",
        boxpoints="outliers",   # muestra outliers como puntos
        showlegend=False,
        boxmean="sd"
    ),
    row=2, col=1
)

fig.add_trace(
    go.Box(
        x=durations_filtered,
        name="Duration",
        marker_color="forestgreen",
        line_color="forestgreen",
        boxpoints="outliers",   # muestra outliers como puntos
        showlegend=False,
        boxmean="sd"
    ),
    row=3, col=1
)

# --- 4. Layout ---
fig.update_layout(
    template="plotly_white",
    title_text="Duration distribution",
    font=dict(size=15),
    title_font=dict(size=20),
    bargap=0.02,
    barmode="overlay",
)
fig.update_yaxes(title_text="Count", row=1, col=1)
fig.update_yaxes(showticklabels=False, row=2, col=1)
fig.update_xaxes(title_text="Duration (s)", row=2, col=1)

fig.show()
fig.write_image("assets/duration_distribution.png", scale=5)
```

### 3.5. Label distribution

```python
matrix = pd.crosstab(df['Wheeze'], df['Crackle'])
matrix = matrix.reindex(index=[0, 1], columns=[0, 1], fill_value=0)

z = matrix.values
total = z.sum()

# Texto personalizado: count + porcentaje
annotations_text = [
    [f'{z[i][j]}<br>({z[i][j]/total*100:.1f}%)' for j in range(z.shape[1])]
    for i in range(z.shape[0])
]

x_labels = ['0', '1']
y_labels = ['0', '1']

fig = ff.create_annotated_heatmap(
    z=z,
    x=x_labels,
    y=y_labels,
    annotation_text=annotations_text,
    colorscale='BuPu',
    showscale=True
)

fig.update_layout(
    title=dict(
        text='Contingence matrix: Wheeze vs Crackle',
        y=0.98,          # posición vertical del título (más cerca del borde superior)
        x=0.5,
        xanchor='center'
    ),
    xaxis_title='Crackle',
    yaxis_title='Wheeze',
    margin=dict(t=100),  # aumenta el espacio superior total (por defecto suele ser ~60-80)
    width=500,
    height=500,
    font=dict(size=14)
)
fig.update_yaxes(autorange="reversed")  # opcional, orden clásico de matriz

fig.show()
fig.write_image("assets/label_distribution.png", scale=5)
```
